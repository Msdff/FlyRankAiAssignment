```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd, numpy as np
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# The label: a page is 'declining' when its recent trend is down. Simple, honest starter label.
df["is_declining_label"] = df["trend_direction"].str.lower().eq("down").astype(int)
print(df.shape[0], "pages |  declining rate:", round(df["is_declining_label"].mean(), 3))
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Discovery 1: how many old pages are still getting real traffic?
old_but_active = df[(df["content_age_days"] >= 365) & (df["pageviews_90d"] >= 100)]
print("Old pages (365+ days) still getting 100+ pageviews:", len(old_but_active), "out of", len(df))

# Discovery 2: does content length relate to how many sessions a page gets?
short_pages = df[df["word_count"] < 1200]
long_pages = df[df["word_count"] >= 1200]
print("Avg sessions_90d — short pages (<1200 words):", round(short_pages["sessions_90d"].mean(), 1))
print("Avg sessions_90d — long pages (1200+ words):", round(long_pages["sessions_90d"].mean(), 1))

# Discovery 3: how many pages get any traffic from AI tools?
ai_referred = df[df["ai_sessions_90d"] > 0]
print("Pages with at least 1 AI-referred session:", len(ai_referred), "out of", len(df))

# Discovery 4: does competition level relate to click-through rate?
print(df.groupby("competition_level")["ctr"].mean().round(2))

```

**What this shows:**

Out of 30,000 pages in the starter dataset, 54.2% currently show a declining trend confirming that decline is a common, not rare, pattern worth prioritizing.

365 pages are both old (365+ days) and still generating meaningful traffic (100+ pageviews in the last 90 days) these are strong refresh candidates, since real audience attention already exists on them even as they age.

Content length shows a striking relationship with engagement: short pages (under 1,200 words) average just 3.3 sessions over 90 days, while longer pages (1,200+ words) average 42.2 sessions — over 12x higher. This suggests word count could be a meaningful signal for prioritization, though it may also reflect deeper factors like topic type or search intent that are worth investigating further.

Only 1,930 pages (about 6.4%) show any AI-referred traffic at all, confirming the lane guide's warning that this signal is sparse — useful for exploratory patterns, but not reliable enough to build a standalone model on.

Click-through rate also varies by competition level: low-competition pages have the highest average CTR (0.34), while medium-competition pages have the lowest (0.24) — suggesting competition context matters when judging whether a page's CTR is actually underperforming or normal for its category.

Together, these numbers confirm multiple independent signals exist in the starter dataset decline rate, content depth, traffic longevity, and competition context that a priority score could meaningfully combine over the next seven weeks.
